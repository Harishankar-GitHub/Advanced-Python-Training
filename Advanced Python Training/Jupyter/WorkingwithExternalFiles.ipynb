# Working with External Files. Also Group and Aggregate Data

```python
import numpy as np
import pandas as pd
```

```python
# Free datasets website - https://www.kaggle.com/datasets

# we may load external files from Excel, csv, json etc.
df = pd.read_csv('data/salaries.csv') # also Excel, json, xml etc.
df.describe()
```

```python
df.columns
df.size # total number of data points
df.sum()
df.max() # or min(), count() etc.
df['phd']
df.sex
# df['rank'] # careful - rank is a keyword
```

```python
# Grouping Data
df_rank = df.groupby(['rank'])
df_rank # we have a group object
df_rank.count()
```

```python
df_n = df.groupby(['salary','sex']) # careful - the data may be extensive
df_n.max()
```

```python
# find the mean salary and service for each rank
l = ['salary', 'service']
gb = df.groupby('rank')
gb[l].mean()
```

```python
# mini-challenge
# show the mean salary fpr each rank by male/female
# find mean salary where salary exceeds 120000

# Solution
a = df.groupby(['rank', 'sex'])
a[['salary']].mean()
```

# Aggregate Data

```python
df.agg({'rank' : ['nunique']})  # number of unique occurrances
df.agg({'sex' : ['nunique']})   # number of unique occurrances
```

```python
df[['phd', 'salary']]
df[['phd', 'salary']].agg(['min', 'max', 'mean'])
```

```python
# This code and the above code are same.

stats=['min', 'max', 'mean']
which = ['phd', 'salary']

i = df[which]
r = i.agg(stats)
r
```
