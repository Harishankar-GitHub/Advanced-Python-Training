## Flights
### Data Analysis of Flights.csv

```python
import numpy as np
import pandas as pd
```

#### Create a DataFrame from CSV data

```python
df = pd.read_csv('data/flights.csv')
df
```

#### Describe

```python
df.describe()
```

#### List the data types of each columns

```python
df.dtypes
```

#### List all Columns

```python
df.columns
```

#### Size of the data

```python
df.size
```

#### Mean flight time

```python
df_by_carrier = df.groupby('carrier')
df_by_carrier['air_time'].mean()
```

#### Group by US Carrier

```python
df_US = df[df['carrier'] == 'US']
df_US
```

#### Group by AS Carrier

```python
df_AS = df[df['carrier'] == 'AS']
df_AS
```

#### Number of flight records

```python
df['flight'].count()
```

#### min, mean, and max delays for departure and arrival

```python
df[['dep_delay','arr_delay', 'distance']].agg(['min','mean','max'])
```

#### Number of unique airlines

```python
grouped_by_carrier = df.groupby('carrier')
len(grouped_by_carrier)
```

#### Number of unique aircrafts represented (Tail Number)

```python
grouped_by_tailnum = df.groupby('tailnum')
print(len(grouped_by_tailnum))

# OR

print(df['tailnum'].nunique())
# OR

print(df.tailnum.nunique())
```

#### Minimum departure delay for your chosen airlines (or all)

```python
grouped_by_carrier = df.groupby('carrier')
grouped_by_carrier['dep_delay'].min()
```

#### Mean departure delay for your chosen airlines (or all)

```python
grouped_by_carrier['dep_delay'].mean()
```

#### Greatest recorded delay

```python
grouped_by_carrier['dep_delay'].max()
```

#### Mean delay for the first 50 records

```python
first_50_records = df.head(50)
first_50_records

grouped_by_carrier = first_50_records.groupby('carrier')
grouped_by_carrier['dep_delay'].mean()
```
